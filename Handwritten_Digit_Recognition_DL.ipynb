```python
import tensorflow as tf
import matplotlib.pyplot as plt

# Load the handwritten digits dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Dataset Loaded Successfully!")
print("Training images:", x_train.shape)
print("Testing images:", x_test.shape)
```

```python
plt.figure(figsize=(10,4))

for i in range(10):
  plt.subplot(2,5,i+1)
  plt.imshow(x_train[i],cmap="gray")
  plt.title("Digit: " +str(y_train[i]))
  plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Normalize image pixel values
x_train = x_train / 255.0
x_test = x_test / 255.0

print("Images normalized successfully!")
print("Training data shape:", x_train.shape)
print("Testin data shape:", x_test.shape)
```

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(28, 28)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Deep Learning Model Created Successfully!")
```

```python
history= model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.1
)

print("Model Training Completed Successfully!")
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.1
)

print("Model Training Completed Successfully!")
```

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test,)

print("Model Evaluation Completed Successfully!")
print("Test Accuracy:", round(float(test_accuracy) * 100,2),"%")
print("Test Loss:", round(float(test_loss), 4))
```

```python
import matplotlib.pyplot as plt
import numpy as np

index=0

prediction = model.predict(x_test[index:index+1], verbose=0)
predicted_digit = np.argmax(prediction)

plt.imshow(x_test[index], cmap="gray")
plt.title("Predicted Digit: " + str(predicted_digit))
plt.axis("off")
plt.show()

print("Actual Digit:", y_test[index])
print("Predicted Digit:", predicted_digit)
```
